```python
import tensorflow.keras as keras
import numpy as np
import matplotlib.pyplot as plt
from random import randint
import cv2 as cv
%matplotlib inline
```

```python
xTrain = np.load('./tmp/xTrain.npy')
yTrain = np.load('./tmp/yTrain.npy')
xTest = np.load('./tmp/xTest.npy')
yTest = np.load('./tmp/yTest.npy')
```

```python
labelSets = ['apple', 'banana', 'bed', 'book', 'cup', 'door', 'foot', 
             'hamburger', 'key', 'sun']
numClasses = len(labelSets)
```

```python
# 验证准确性
index = randint(0, len(xTrain))
print(labelSets[int(yTrain[index])])
trainSingleImage = xTrain[index].reshape(28, 28)
cv.imwrite('./tmp/test1.png', trainSingleImage)
plt.imshow(trainSingleImage, cmap='gray')
plt.show()

print('-------------------')

index = randint(0, len(xTest))
print(labelSets[int(yTest[index])])
testSingleImage = xTest[index].reshape(28, 28)
cv.imwrite('./tmp/test2.png', testSingleImage)
plt.imshow(testSingleImage, cmap='gray')
plt.show()
```

```python
# 数据集是经过展平化的图像，为了方便使用卷积层，需要将图像shape还原为 [28, 28, 1]
xTrain = xTrain.reshape(xTrain.shape[0],28,28,1).astype('float32')
xTest = xTest.reshape(xTest.shape[0],28,28,1).astype('float32')
```

```python
# 数据预处理把像素压缩到 0~1
xTrain /= 255.0
xTest /= 255.0
```

```python
# 独热编码操作
yTrain = keras.utils.to_categorical(yTrain,numClasses)
yTest = keras.utils.to_categorical(yTest,numClasses)
```
